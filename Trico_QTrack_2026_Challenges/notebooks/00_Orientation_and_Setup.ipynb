## First circuit

Run this example and observe the difference between exact theoretical probabilities and simulated measurement counts. The leftmost character of a bitstring is the highest-numbered qubit.

```python
qc = QuantumCircuit(1)
qc.h(0)
print('Circuit:')
print(qc.draw(output='text'))
print('Exact probabilities:',Statevector.from_instruction(qc).probabilities_dict())
qc.measure_all()
counts = dict(AerSimulator().run(qc,shots=256,seed_simulator=42).result().get_counts())
print('256 simulated measurements:',counts)
assert sum(counts.values()) == 256
```
